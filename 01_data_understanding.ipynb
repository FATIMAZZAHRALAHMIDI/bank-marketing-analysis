# Is the dataset balanced or imbalanced? Why does it matter? 
**Answer:** No, the dataset is highly **imbalanced**. 
About 88.3% of clients said "no", and only 11.7% said "yes".

**Why is it important ?**
If we train a model on this dataset, a "lazy" model that always predicts "no" would have an 88% accuracy. However, it would be completely useless for the bank because it would never identify the clients who say "yes". 

We will need to handle this imbalance later by using specific metrics (like Recall and F1-score) instead of just Accuracy, and by applying techniques like class weighting.

**Group E: Synthesis** 
### Data Understanding Summary

**Dataset Overview:** 
The dataset contains 45,211 clients and 17 variables (7 numerical, 10 categorical). There are no duplicate rows and no explicit missing values (NaN). 

**Target Variable (`y`):** 
The dataset is highly **imbalanced**: 88.3% of clients said "no" and only 11.7% said "yes". This will be a major challenge for Machine Learning.

**Data Quality Issues:** 
We found hidden missing values represented as `"unknown"`, especially in `poutcome` (81.7%) and `contact` (28.8%). We also found that `pdays = -1` means the client was never contacted before. 

**Next Steps:** 
We need to find a strategy to handle the `"unknown"` values in Phase 2 (Cleaning), and we must be careful with the class imbalance when building our prediction model.

# 04 — Statistical Analysis

## Bank Marketing Dataset

**Goal:** Use statistical tests to prove if the relationships observed in the EDA are mathematically significant (or just due to chance).

**Tools used:** Chi-square test (for categorical variables), T-test (for numerical variables).

**Date:** 2026-10-25

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency, ttest_ind

sns.set_style("whitegrid")
pd.set_option('display.max_columns', None)


df = pd.read_csv('../data/bank_clean.csv')

df['y_numeric'] = df['y'].map({'yes': 1, 'no': 0})

print("Data loaded and ready for stats ✅")
```

```python
# Is the relationship between profession and subscription significant? (Chi² test)
table = pd.crosstab(df['job'],df['y'])
chi2, p_value, dof, expected = chi2_contingency(table)
print(f"Chi-square statistic: {chi2:.2f}")
print(f"P-value: {p_value:.5e}") 
print()
if p_value < 0.05 : 
    print('The relationship is statistically significant. Job influences subscription')
else:
    print('Not statistically significant')
```

```python
# Is the relationship between education and subscription significant? (Chi²) 
table1= pd.crosstab(df['education'] , df['y']) 
chi2 , p_value , dof ,expected = chi2_contingency(table1) 
print(f'Chi_squaer :{chi2:.2f}') 
print(f'P_value : {p_value:.5e}')
if p_value < 0.05 : 
    print('The relationship is statistically significant. education influences subscription')
else :
   print ('Not statistically significant') 
```

```python
# Is the relationship between housing loan and subscription significant? (Chi²) 
table2 = pd.crosstab(df['housing'] , df['y']) 
chi2 , p_value , dof ,expected = chi2_contingency(table2) 
print(f'Chi_squaer :{chi2:.2f}') 
print(f'P_value : {p_value:.5e}')
if p_value < 0.05 : 
    print('The relationship is statistically significant. housing influences subscription')
else :
   print ('Not statistically significant') 
```

```python
#  Does the mean age differ between yes and no clients? (t-test) 
yes_age = df[df['y'] == 'yes']['age']
no_age = df[df['y'] =='no']['age']
t_stat, p_value = ttest_ind(yes_age , no_age)  
print(f'mean_age (yes): {yes_age.mean():.2f}')
print(f'mean_age (no): {no_age.mean():.2f}')
print(f'P-value: {p_value:.5e}')
if p_value < 0.05 :
   print ('The age difference is statistically significant')
else : 
   print('The age difference is NOT statistically significant')

```

```python
# Does the mean balance differ between yes and no? (t-test) 
balance_yes = df[df['y'] == 'yes']['balance']
balance_no = df[df['y'] == 'no']['balance'] 
t_test , p_value = ttest_ind(balance_yes , balance_no)
print(f'mean balance(yes) : {balance_yes.mean() :.2f}')
print(f'mean balance(no) : {balance_no.mean() :.2f}')
print(f'p_value : {p_value :.5e}')

if p_value < 0.05 : 
    print('The balance difference is statistically significant')
else : 
    print('The balance difference is NOT statistically significant')
```

```python
# Does the mean number of calls differ between yes and no? (t-test) 
Campaign_yes = df[df['y'] == 'yes']['campaign']
Campaign_no = df[df['y'] == 'no']['campaign'] 
t_test , p_value = ttest_ind(Campaign_yes , Campaign_no)
print(f'mean Campaign(yes) : {Campaign_yes.mean() :.2f}')
print(f'mean Campaign(no) : {Campaign_no.mean() :.2f}')
print(f'p_value : {p_value :.5e}')

if p_value < 0.05 : 
    print('The Campaign difference is statistically significant')
else : 
    print('The Campaign difference is NOT statistically significant') 
```

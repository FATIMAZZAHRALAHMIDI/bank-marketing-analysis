# 07 — Business Recommendations

## Bank Marketing Campaign Strategy

**Goal:** Translate the data analysis and machine learning results into actionable recommendations for the bank's marketing team.


## Priority Target Profile

Based on the EDA and the model, the bank should prioritize the following profiles:
- **Profession:** Students and Retired people (highest conversion rates).
- **Age:** Young adults (18-30) or Seniors (60+).
- **Balance:** Clients with a high bank balance (especially > 60,000€).
- **Loan Status:** Clients with NO housing loan and NO personal loan.
- **History:** Clients who had a "success" in a previous campaign (this is the strongest signal).
- **Timing:** Contact them in March, September, October, or December.

**Business Action:** Create a "VIP list" combining these criteria and call them first.

## Clients to Avoid

The bank should avoid spending time and money on these profiles, as they have a very low probability of subscribing:
- **Profession:** Blue-collar workers.
- **Age:** 30 to 40 years old (the lowest conversion rate).
- **Balance:** Low or negative balance (near 0€).
- **Debt:** Clients with a housing loan AND/OR a personal loan.
- **Contact History:** Clients who have been called more than 5 times (campaign fatigue).
- **Timing:** Avoid calling in May (the worst month for conversion).

**Business Action:** Do not include these clients in the priority calling list to save time and reduce costs.


### Estimated Gain in Conversion Rate

Currently, the baseline conversion rate is 11.7%. 

Thanks to the targeted profiles identified by the EDA and the Random Forest model, the bank can expect a much higher conversion rate. Since our model successfully identifies 39% of the actual subscribers (Recall), and our EDA showed that specific profiles (like students or high-balance clients) have a conversion rate over 30%, we can estimate that **calling the right people will result in approximately 30 to 39 subscriptions for every 100 calls made**. 

This is a massive improvement compared to the current 11.7%, saving time and reducing campaign costs.

## Top 3 Business Recommendations

1. **Focus on the "Golden Profiles":** Stop calling everyone. Prioritize students, retired people, and clients with a balance over 60,000€ who have no loans. This will drastically increase the conversion rate.
2. **Limit the Number of Calls:** Our analysis proved that calling a client more than 5 times decreases their chances of subscribing (campaign fatigue). The bank should limit calls to a maximum of 3 attempts per client.
3. **Time the Campaigns Right:** Avoid May and summer months. Concentrate 80% of the marketing budget on March, September, October, and December, which have the highest conversion rates.

## Limitations of the Analysis

- **Historical Data:** The dataset is from an old campaign. Client behaviors may have changed since then.
- **Missing External Factors:** The dataset does not include external factors like the economic situation, competitor offers, or the specific sales agent's skills.
- **The "Duration" Bias:** Call duration is the most important feature, but it is only known *during* the call. The bank cannot know in advance how long a client will stay on the phone.

```python
# Core libraries 
import pandas as pd 
import numpy as np 
import math as m 

# Visualization 
import matplotlib.pyplot as plt 
import seaborn as sns 

# Display options 
pd.set_option('display.max_columns', None) 
sns.set_style('whitegrid')  

print("Libraries imported successfully!") 
```

```python
#Load Data 
df = pd.read_csv('../data/bank-full.csv', sep=';')
print("Data loaded successfully!") 
```

```python
# remove duplicates 
df.duplicated().sum()
# 0 duplicates 
```

#Handling "unknown" values

We decided to **keep** the "unknown" values as a separate category. 
Deleting them would mean losing over 80% of our dataset (especially because of `poutcome`). 
For the Machine Learning model, "unknown" is a valid information (e.g., the client was never contacted before).


```python
# Are there outliers in balance? In campaign? How do you see them? 

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Graphique 1 : Balance 
sns.boxplot(x=df['balance'], ax=axes[0], color="skyblue")
axes[0].set_title('Outliers in Balance')

# Graphique 2 : Campaign
sns.boxplot(x=df['campaign'], ax=axes[1], color="lightgreen")
axes[1].set_title('Outliers in Campaign')


plt.show()
```

# Handling outliers

We decided to **keep** the outliers in the dataset. 
As seen in the boxplots, the outliers in `balance` represent clients with very high amounts of money, and the outliers in `campaign` represent clients who were called many times. 

These are not data errors; they are real business cases (VIP clients or aggressive marketing campaigns). Removing them would mean losing valuable information. However, during the Machine Learning phase, we will use models that are less sensitive to outliers (like Random Forest).

```python
df.to_csv('../data/bank_clean.csv', index=False)
print("Clean dataset saved!")
```
